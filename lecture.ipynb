```python
df = pd.read_csv("./dataset/diamonds.csv")
df.head(2)
```

```python
df.groupby("cut")["depth"].max()
# 대상 변수가 2개 이상이 아니면 시리즈로 나옴
```

```python
df_g1 = df.groupby("cut")["depth"].max().reset_index()
df_g1
```

```python
df.groupby("cut")[["depth", "x"]].max()
```

```python
series_g2 = df.groupby(["cut", "color"])["x"].max()
series_g2.head()
# 얘는 기준 변수가 2개라서 시리즈로 나와
```

```python
df_g2 = df.groupby(["cut", "color"])["x"].max().reset_index()
df_g2.head()
```

```python
df_g3 = df.groupby("cut")["y"].agg(["min", "max"]).reset_index()
df_g3
```

```python
df_g4 = df.groupby("cut")[["x", "y"]].agg({"x": "min", "y": "max"}).reset_index()
df_g4
```

```python
df = pd.read_csv("./dataset/elec_load_2017_7d.csv")
df.head(2)
```

```python
df_m1 = df.melt(id_vars = ["YEAR", "MONTH", "DAY"])
df_m1
```

```python
df_m2 = df.melt(id_vars = ["YEAR", "MONTH", "DAY"],
               var_name="HOUR", value_name="load")
df_m2
```

```python
df_m2["h1"] = df_m2["HOUR"].str.replace("X", "").str.replace("HR", "")
df_m2["h2"] = df_m2["HOUR"].str.replace("X|HR", "", regex=True)
df_m2["h3"] = df_m2["HOUR"].str.replace("[A-Z]", "", regex=True)
df_m2["h4"] = df_m2["HOUR"].str.replace("[^0-9]", "", regex=True)
df_m2
```
