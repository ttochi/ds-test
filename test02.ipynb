```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
```

```python
df = pd.read_csv("test/set_02_data.csv")
df.head()
```

```python
df_q1 = df.loc[(df["gender"] == "Male") & (df["SeniorCitizen"] == 0), ].reset_index(drop = True)
len(df_q1)
```

```python
df_q1 = df_q1[["OnlineSecurity", "OnlineBackup", "DeviceProtection", "TechSupport", "StreamingTV", "StreamingMovies"]].copy()

def yes_counter(x):
    if not x.isin(["Yes", "No"]).all():
        return np.nan
    return (x == "Yes").sum()

df_q1["yes_cnt"] = df_q1.apply(yes_counter, axis = 1)

df_q1 = df_q1.dropna()

df_q1["yes_cnt"].value_counts()
```
