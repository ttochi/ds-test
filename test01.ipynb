```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_csv("./test/set_01_data.csv")
df.head(5)
```

```python
df["F1"] = np.where(df["Frequency"].str.contains("Hz"), df["Frequency"].str.replace("[^0-9]", "", regex=True), np.nan)
df["F2"] = np.where(df["Picture_quality"].str.contains("Hz"), df["Picture_quality"].str.replace("[^0-9]", "", regex=True), np.nan)
df["F3"] = np.where(df["Speaker"].str.contains("Hz"), df["Speaker"].str.replace("[^0-9]", "", regex=True), np.nan)
df.head(5)
```

```python
df["Freq"] = np.where(df["F1"].isna(), np.where(df["F2"].isna(), df["F3"], df["F2"]), df["F1"])
df.head(5)
```

```python
df_60 = df.loc[df["Freq"] == "60", ]
len(df_60)
```

```python
# Question 2
df = pd.read_csv("./test/set_01_data.csv")

df_sub = df.loc[df["Picture_quality"].str.contains("4K") | df["Operating_system"].str.contains("4K") | df["channel"].str.contains("4K"), ]
df_sub = df_sub.loc[(df_sub["Stars"] != 0) & (df_sub["Ratings"] != 0) & (df_sub["Reviews"] != 0), ]
df_sub = df_sub.loc[:, ["Stars", "Ratings", "Reviews", "current_price", "MRP"]].reset_index(drop = True)
df_sub.head()
```

```python
kmeans = KMeans(n_clusters=3, random_state=123).fit(df_sub)
```

```python
kmeans.labels_
```

```python
df_sub["cluster"] = kmeans.labels_
df_sub["cluster"].value_counts()
```

```python
# Question 3
df = pd.read_csv("./test/set_01_data.csv")
df.head(5)
```

```python
df = df.loc[(~df["channel"].str.contains("Oper|Pixel")), ]
df["ReviewRate"] = df["Reviews"] / df["Ratings"]
df["Discount"] = df["current_price"] / df["MRP"]
# df["Netflix"] = np.where(df["channel"].str.contains("Netflix"), 1, 0)
df["Netflix"] = df["channel"].str.contains("Netflix") + 0
# df["Prime"] = np.where(df["channel"].str.contains("Prime"), 1, 0)
df["Netflix"] = df["channel"].str.contains("Prime") + 0
# df["Quality"] = np.where(df["Picture_quality"].str.contains("4K|8K"), 1, 0)
df["Quality"] = df["Picture_quality"].str.contains("4K|8K") + 0

df = df.dropna() # 이거도 뒤에서 컬럼 필터링 한 다음에 해주는게 좋을 듯!!!
len(df)
```

```python
# df_3 = df.loc[:, ["ReviewRate", "MRP", "Discount", "Netflix", "Prime", "Quality"]].reset_index(drop = True)
df_3 = df[["Stars", "ReviewRate", "MRP", "Discount", "Netflix", "Prime", "Quality"]].reset_index(drop = True)
# reset은 위에서 필터링 할 때 미리 해주면 좋을 듯

df_3.head()
```

```python
random = RandomForestRegressor(random_state=123).fit(df_3.drop(columns = "Stars"), df_3[["Stars"]])
```

```python
arr_MDI = random.feature_importances_
arr_MDI
```
